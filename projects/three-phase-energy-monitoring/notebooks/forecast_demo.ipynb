# Three-phase monitoring demo
The dataset is synthetic. The baseline predicts the next hourly reading using the mean of the last three total power observations.

```python
import csv
from pathlib import Path
from statistics import mean

source = Path("examples/three_phase_readings.csv")
if not source.exists():
    source = Path("../examples/three_phase_readings.csv")
with source.open(newline="") as file:
    readings = list(csv.DictReader(file))
totals = [sum(float(row[phase]) for phase in ("l1_kw", "l2_kw", "l3_kw")) for row in readings]
prediction = mean(totals[-3:])
print(f"Next-hour baseline forecast: {prediction:.2f} kW")
```
